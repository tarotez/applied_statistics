```python
import matplotlib.pyplot as plt
import numpy as np
from numpy import array as ar
import pandas as pd

df = pd.read_csv('Diabetes.csv')
diabetes = df['Target']
cholesterol = df['Cholesterol']
LDL = df['LDL']
BP = df['BP']
BMI = df['BMI']

x1g, x2g, x3g, x4g, yg = cholesterol, LDL, BP, BMI, diabetes

def prediction(x,w):
    return x @ w
```

```python
X = np.array([x1g, np.ones(len(x1g))]).T
w = np.linalg.solve(X.T @ X, X.T @ yg)
error = yg - prediction(X,w)
RMSE = float(np.sqrt(error.T @ error / len(error)))
print('RMSE =', RMSE)
```

```python
plt.scatter(x1g, yg, color='blue')
plt.scatter(x1g, prediction(X, w), color='red')
```
