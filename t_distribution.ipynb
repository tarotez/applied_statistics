```python
# t-distribution
import numpy as np
from numpy.random import normal
from scipy.stats import t, norm

# simulation
mu, sigma = 10, 20
n = 10   # number of samples
df = n - 1 # degree of freedom
trialNum = 10000
tL = []
for _ in range(trialNum):
    x1, x2 = normal(mu, sigma, n), normal(mu, sigma, n)
    avg1 = np.average(x1)
    avg2 = np.average(x2)
    std1 = np.std(x1, ddof=1) # delta degree-of-freedom is set to one
    std2 = np.std(x2, ddof=1) # delta degree-of-freedom is set to one
    std = np.sqrt((std1 ** 2 + std2 ** 2))  # The 2-norm of the standard deviation vector
    standard_error = std / np.sqrt(n)
    t_statistic = (avg1 - avg2) / standard_error
    tL.append(t_statistic)

# analytic t and normal
xmin, xmax = np.percentile(tL, 0.01), np.percentile(tL, 99.99)
segmentNum = 100
x = np.linspace(xmin, xmax, segmentNum)
t_pdf = t.pdf(x, df, 0, 1)
normal_pdf = norm.pdf(x, 0, 1)
```

```python
%matplotlib inline
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
binNum = 100
ax.hist(tL, binNum)
ax.plot(x, t_pdf * trialNum * (xmax - xmin) / binNum, label='t')
ax.plot(x, normal_pdf * trialNum * (xmax - xmin) / binNum, label='normal')
ax.set_xlabel('t-statistic')
ax.set_ylabel('frequency')
ax.legend()
ax.set_xlim([0,5])
plt.savefig('../figs/t_distribution')
plt.show()
```
