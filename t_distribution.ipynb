```python
# t-distribution
import numpy as np
from numpy.random import normal
from scipy.stats import t, norm

# simulation
mu_1, sigma = 10, 20
mu_2 = mu_1   # null-hypothesis
n = 5   # number of samples
df = n - 1 # degree of freedom
trialNum = 10000
tL = []
for _ in range(trialNum):
    samples_1 = normal(mu_1, sigma, n)
    avg_1 = np.average(samples_1)
    samples_2 = normal(mu_2, sigma, n)
    avg_2 = np.average(samples_2)    
    sample_var_1 = np.sum((samples_1 - avg_1) ** 2) / (n - 1)
    t_statistic = (avg_1 - avg_2) / np.sqrt(sample_var_1 / n)
    tL.append(t_statistic)

# analytic t and normal
xmin, xmax = np.percentile(tL, 0.01), np.percentile(tL, 99.99)
segmentNum = 100
x = np.linspace(xmin, xmax, segmentNum)
t_pdf = t.pdf(x, df, 0, 1)
normal_pdf = norm.pdf(x, 0, 1)
```

```python
%matplotlib inline
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
binNum = 100
ax.hist(tL, binNum)
ax.plot(x, t_pdf * trialNum * (xmax - xmin) / binNum, label='t')
ax.plot(x, normal_pdf * trialNum * (xmax - xmin) / binNum, label='normal')
ax.set_xlabel('(avg(X_1) - avg(X_2)) / sqrt(sv(X_1) / n)')
ax.set_ylabel('frequency')
ax.legend()
ax.set_xlim([-10,10])
plt.savefig('../figs/t_distribution')
plt.show()
```

```python
# Type-II error
d = 0.1
params1 = 5, 1
params2 = params1[0] + d, 1
trialNum = 500
alpha = 0.01

for n in range(0, 5000, 500):
    false_acceptance_cnt = 0
    for _ in range(trialNum):
        pval = ttest_ind([normal(*params1) for _ in range(n)], [normal(*params2) for _ in range(n)]).pvalue 
        if pval > alpha:
            false_acceptance_cnt += 1
    typeIIerror = false_acceptance_cnt / trialNum
    print(typeIIerror)
```
