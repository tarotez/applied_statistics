```python
import pandas as pd
names = []
dfL = []
for name in names:
    df = pd.read_csv('phys_' + name + '.csv')
    df['name'] = [name] * len(df)
    dfL.append(df)
df = pd.concat(dfL, ignore_index=True)
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
for name, data in df.groupby('name'):
    ax.scatter(data['sleep'], data['pulse'], label=name)
ax.legend()
ax.set_xlabel('sleep (H)', fontsize=15)
ax.set_ylabel('pulse', fontsize=15)
ax.tick_params(axis='both', labelsize=15)
plt.show()
```

```python
# biased estimator of variance
import numpy as np
true_mu, true_v, n = 0, 20, 3
estimNum = 100
est_vL = []
for _ in range(estimNum):
    x = np.random.normal(loc=true_mu, scale=np.sqrt(true_v), size=n)
    est_vL.append(np.sum(((x - np.mean(x)) ** 2)) / n)
np.mean(est_vL)
```

```python
# unbiased estimator of variance
est_vL = []
for _ in range(estimNum):
    x = np.random.normal(loc=true_mu, scale=np.sqrt(true_v), size=n)
    est_vL.append(np.sum(((x - np.mean(x)) ** 2)) / (n - 1))
np.mean(est_vL)
```
