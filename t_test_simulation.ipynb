```python
import numpy as np
from numpy.random import normal
import scipy as sp
from scipy.stats import ttest_ind
import matplotlib.pyplot as plt

# Type-I error
params1 = 5, 1
params2 = params1
trialNum = 1000
alpha = 0.05

for n in range(0, 2000, 500):
    false_rejection_cnt = 0
    for _ in range(trialNum):
        pval = ttest_ind([normal(*params1) for _ in range(n)], [normal(*params2) for _ in range(n)]).pvalue 
        if pval < alpha:
            false_rejection_cnt += 1
    typeIerror = false_rejection_cnt / trialNum
    print(typeIerror)
```

```python
# Type-II error
d = 1
params1 = 5, 1
params2 = params1[0] + d, 1
trialNum = 1000
alpha = 0.05

for n in range(0, 2000, 500):
    false_acceptance_cnt = 0
    for _ in range(trialNum):
        pval = ttest_ind([normal(*params1) for _ in range(n)], [normal(*params2) for _ in range(n)]).pvalue 
        if pval > alpha:
            false_acceptance_cnt += 1
    typeIIerror = false_acceptance_cnt / trialNum
    print(typeIIerror)
```
