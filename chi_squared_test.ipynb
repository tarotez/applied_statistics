```python
# bank-additional-full datast
attr = 'marital'
# attr = 'education'
# attr = 'contact'
# attr = 'month'
# attr = 'age'
# attr = 'day_of_week'
# attr = 'default'

# Employee dataset
# attr = 'Education'
# attr = 'Gender'
# attr = 'EverBenched'
contingency_table = pd.crosstab(df[attr], df['default'])
print(contingency_table)
```

```python
chi2, p, dof, expected = chi2_contingency(contingency_table)

print(f"Chi-Square Statistic: {chi2}")
print(f"p-value: {p}")
print(f"Degrees of Freedom: {dof}")
print("Expected Frequencies Table:\n", expected)
```
