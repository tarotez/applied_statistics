```python
from scipy.stats import chi2_contingency
import numpy as np
import pandas as pd

n = 100
attr1L, attr2L = [], []
for _ in range(n):
    attr1 = np.random.choice([0,1])
    attr2 = np.random.choice([0,1])    
    # attr2 = 0 if attr1 else 1
    # attr2 = 0 if np.random.choice([0,1]) else attr2
    # attr2 = 0 if np.random.choice([0,1]) and attr1 else attr2
    attr1L.append(attr1)
    attr2L.append(attr2)
contingency_table = pd.crosstab(attr1L, attr2L)
print(contingency_table)
chi2, p, dof, expected = chi2_contingency(contingency_table)
print(p)
```
