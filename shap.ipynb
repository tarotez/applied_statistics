```python
import numpy as np
from sklearn.linear_model import LinearRegression
import shap
import pandas as pd

df = pd.read_csv('../data/happiness/2015.csv')
attrs = ['Economy (GDP per Capita)', 'Family', 'Health (Life Expectancy)', 'Freedom', 'Trust (Government Corruption)', 'Generosity']
X = np.array([df[attr] for attr in attrs]).T
y = np.array(df['Happiness Score'])

model = LinearRegression()
model.fit(X, y)

explainer = shap.LinearExplainer(model, X)
explanation = explainer(X)

shap_values = explanation.values
shap.plots.violin(explanation, feature_names=attrs)
```
