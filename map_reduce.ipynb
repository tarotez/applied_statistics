```python
# for-loop
s = 0
for i in range(1, 6):
    s += i ** 2
s
```

```python
# list comprehension
sum([i ** 2 for i in range(1, 6)])
```

```python
# map-reduce by defining functions
from functools import reduce

def addition(x, y):
    return x + y

def square(x):
    return x ** 2

reduce(addition, map(square, range(1, 6)))
```

```python
# map-reduce by using a pre-defined operator
from functools import reduce
from operator import add 

def square(x):
    return x ** 2

reduce(add, map(square, range(1, 6)))
```

```python
# map-reduce with anonymous function
from functools import reduce
reduce(lambda cmtv, x: cmtv + x, map(lambda x: x ** 2, range(6)))
```

```python
# inner product using map-reduce
from functools import reduce
from operator import add

a = [1, 2, 3]
b = [3, 2, 1]
reduce(add, map(lambda x: x[0] * x[1], zip(a, b)))
```
