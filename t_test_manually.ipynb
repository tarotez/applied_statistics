```python
import numpy as np
from scipy.stats import ttest_ind
```

```python
d1, d2 = [1, 2, 3], [0, 2]
t, p = ttest_ind(d1, d2)
float(t)
```

```python
# manual
n1, n2 = len(d1), len(d2)
df1, df2 = n1 - 1, n2 - 1
avg1, avg2 = np.average(d1), np.average(d2)
std1, std2 = np.std(d1, ddof=1), np.std(d2, ddof=1) # delta degree-of-freedom is set to one
pooled_std = np.sqrt((df1 * std1 ** 2 + df2 * std2 ** 2) / (df1 + df2))
t_manual = (avg1 - avg2) / (pooled_std * np.sqrt(1/n1 + 1/n2))
float(t_manual)
```
