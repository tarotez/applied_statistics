```python
import numpy as np
from scipy.stats import ttest_ind
```

```python
d1, d2 = [1, 2, 3], [0, 1, 2]
t, p = ttest_ind(d1, d2)
float(t)
```

```python
# manual
n = len(d1)
avg1 = np.average(d1)
avg2 = np.average(d2)
std1 = np.std(d1, ddof=1) # delta degree-of-freedom is set to one
std2 = np.std(d2, ddof=1) # delta degree-of-freedom is set to one
std = np.sqrt((std1 ** 2 + std2 ** 2))  # The 2-norm of the standard deviation vector
standard_error = std / np.sqrt(n)
t_manual = (avg1 - avg2) / standard_error
float(t_manual)
```
