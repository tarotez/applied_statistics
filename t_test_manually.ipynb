```python
import numpy as np
from scipy.stats import ttest_ind
```

```python
d1, d2 = [1, 2, 3], [0, 2]
t, p = ttest_ind(d1, d2)
float(t)
```

```python
# manual
n1, n2 = len(d1), len(d2)
df1, df2 = n1 - 1, n2 - 1
avg1, avg2 = np.average(d1), np.average(d2)
var1, var2 = np.var(d1, ddof=1), np.var(d2, ddof=1)   # delta degree-of-freedom is set to one
pooled_var = (df1 * var1 + df2 * var2) / (df1 + df2)
n_harmonic = 1 / (1/n1 + 1/n2)
standard_error = np.sqrt(pooled_var / n_harmonic)
t_manual = (avg1 - avg2) / standard_error
#float(t_manual)
```
