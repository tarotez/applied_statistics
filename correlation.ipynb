```python
import numpy as np
import pandas as pd
names = []
avg_by_name = {name:{} for name in names}
for experiment in ['phys', 'sleep']:
    for name in names:
        df = pd.read_csv(experiment + '_' + name + '.csv')
        for attr in df.columns:
            avg_by_name[name][attr] = np.nanmean(df[attr])
```

```python
import matplotlib.pyplot as plt
attrL = ['sys', 'dia', 'sleep', 'pulse']
fig, axM = plt.subplots(len(attrL), len(attrL))
fig.set_figheight(12)
fig.set_figwidth(12)
def dev(vec):
    return vec - np.mean(vec)
show_legend = True
for attr1, axL in zip(attrL, axM):
    for attr2, ax in zip(attrL, axL):
        x, y = [], []
        for i, name in enumerate(names):
            avg = avg_by_name[name]
            ax.scatter(avg[attr1], avg[attr2], label=name, color=plt.cm.tab20b(i))
            x.append(avg[attr1])
            y.append(avg[attr2])
        print(attr1 + ', ' + attr2)
        print(np.corrcoef(x,y, bias=True)[0,1])
        ax.set_xlabel(attr1)
        ax.set_ylabel(attr2)
        if show_legend:
            ax.legend(fontsize=4)
            show_legend = False
plt.savefig('scatter_phys_sleep.png')
plt.show()
```
