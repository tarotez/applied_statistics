```python
import numpy as np
import pandas as pd
names = ["Reo Kobayashi", "Liu", "honoka", "mihiro", "taisei", "Kotaro", "tan", "rana", "kei", "rin", "rena", "Kazu"]
avg_by_name = {name:{} for name in names}
for experiment in ['phys', 'sleep']:
    for name in names:
        df = pd.read_csv(experiment + '_' + name + '.csv')
        for attr in df.columns:
            avg_by_name[name][attr] = np.nanmean(df[attr])
```

```python
%matplotlib inline
```

```python
import matplotlib.pyplot as plt
attr1, attr2 = 'dia', 'sys'
fig, ax = plt.subplots()
for i, name in enumerate(names):
    avg = avg_by_name[name]
    ax.scatter(avg[attr1], avg[attr2], label=name, color=plt.cm.tab20(i))
ax.set_xlabel(attr1)
ax.set_ylabel(attr2)
ax.legend()
plt.show()
```
