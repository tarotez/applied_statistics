```python
from scipy.stats import f_oneway
import pandas as pd
import numpy as np
df = pd.read_csv('winequality-red.csv', sep=';')
# attr = 'residual sugar'
# attr = 'pH'
attr = 'density'
```

```python
levels = [5,6,7]
obsD = {}
for level in levels:
    obsD[level] = df[df['quality'] == level][attr]
```

```python
obsD
```

```python
# plot distributions
import matplotlib.pyplot as plt
for level in levels:
    h = plt.hist(obsD[level], bins=30, alpha=0.5)
plt.title(attr)
plt.xlabel(attr)
plt.legend(['Quality ' + str(level) for level in levels])
plt.savefig('wine_' + attr + '.png')
# plt.show()
```

```python
obsL = [obsD[level] for level in levels]
```

```python
stat, p = f_oneway(obsL[0], obsL[1], obsL[2])
print(p)
```

```python
stat, p = f_oneway(*obsL)
print(p)
```
