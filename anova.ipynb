```python
from scipy.stats import f_oneway
import pandas as pd
import numpy as np
df = pd.read_csv('winequality-red.csv', sep=';')
attr = 'density'
# attr = 'residual sugar'
# attr = 'pH'
```

```python
q4 = df[df['quality'] == 4][attr]
q5 = df[df['quality'] == 5][attr]
q6 = df[df['quality'] == 6][attr]
```

```python
stat, p = f_oneway(q4, q5, q6)
float(p)
```
