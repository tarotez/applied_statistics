```python
from scipy.stats import ttest_ind, levene, shapiro
import pandas as pd
df = pd.read_csv('winequality-red.csv', sep=';')
# attr = 'density'
# attr = 'alcohol'
attr = 'residual sugar'
# attr = 'pH'
# attr = 'sulphates'
q5 = df[df['quality'] == 5][attr]
q6 = df[df['quality'] == 6][attr]
```

```python
# plot distributions
import matplotlib.pyplot as plt
h = plt.hist(q5, bins=30, alpha=0.5)
h = plt.hist(q6, bins=30, alpha=0.5)
plt.title(attr)
plt.legend(['Quality 5', 'Quality 6'])
```

```python
p = ttest_ind(q5, q6).pvalue
float(p)
```

```python
p = levene(q5, q6).pvalue
float(p)
```

```python
p = shapiro(q5).pvalue
float(p)
```
