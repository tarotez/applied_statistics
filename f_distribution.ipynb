```python
import numpy as np
from numpy.random import normal
from scipy.stats import f

# simulation
sigma = 10
n = 20   # number of samples
df = n - 1 # degree of freedom
trialNum = 10000
fL = []
for _ in range(trialNum):
    f_statistic = np.var(normal(0, sigma, n)) / np.var(normal(0, sigma, n))
    fL.append(f_statistic)

# analytic
xmin, xmax = np.percentile(fL, 0.01), np.percentile(fL, 99.99)
segmentNum = 100
x = np.linspace(xmin, xmax, segmentNum)
f_pdf = f.pdf(x, df, df, 0, 1)
```

```python
%matplotlib inline
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
binNum = 100
ax.hist(fL, binNum)
ax.plot(x, f_pdf * trialNum * (xmax - xmin) / binNum, label='F')
ax.set_xlabel('var(x_1) / var(x_2)')
ax.set_ylabel('frequency')
ax.legend()
ax.set_xlim([0,5])
plt.savefig('../figs/f_distribution')
plt.show()
```
