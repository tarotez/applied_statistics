```python
# for-loop
s = 0
for _ in range(0, int(1e6)):
    s += 1 ** 2
print('sum =', s)
```

```python
# numpy array
import numpy as np
v = np.ones(int(1e6))
s = v.T @ v
print('sum =', s)

# which is faster, for-loop or multiplication of numpy arrays?
# what is the demerit of using a numpy array for this computation?
```

```python
# numpy array
import numpy as np
v = np.ones(int(1e6))
s = np.linalg.norm(v) ** 2
print('sum =', s)
```
