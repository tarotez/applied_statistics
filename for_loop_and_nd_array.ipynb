```python
# which is faster, for-loop or multiplication of numpy arrays?
import numpy as np
v1 = np.random.random(int(1e7))
v2 = np.random.random(int(1e7))
```

```python
# for-loop
s = 0
for x1, x2 in zip(v1, v2):
    s += x1 * x2
print('sum =', s)
```

```python
# numpy array
import numpy as np
s = v1.T @ v2
print('sum =', s)
```
